            Anomaly Detection
        (using Credit Card Fraud Dataset)

```python
import numpy as np
import pandas as pd
from Anomaly_Detection import *
from sklearn.model_selection import train_test_split
```

```python
#Data-Preprocessing

df=pd.read_csv("creditcard.csv")

normal_data = df[df["Class"] == 0]
fraud_data = df[df["Class"] == 1]

X_normal=normal_data.drop("Class",axis=1).values
X_fraud=fraud_data.drop("Class",axis=1).values

X_train, X_temp = train_test_split(X_normal,test_size=0.4,random_state=42)

X_cv_normal, X_test_normal = train_test_split(X_temp,test_size=0.5,random_state=42)

X_cv = np.vstack([X_cv_normal,X_fraud[:200]])

y_cv = np.hstack([np.zeros(len(X_cv_normal)),np.ones(200)])

X_test = np.vstack([X_test_normal,X_fraud[200:]])

y_test = np.hstack([np.zeros(len(X_test_normal)),np.ones(len(X_fraud[200:]))])
```

```python
#Normalization 
mu = np.mean(X_train, axis=0)
sigma = np.std(X_train, axis=0)
sigma[sigma == 0] = 1e-15

X_train = (X_train - mu) / sigma
X_cv = (X_cv - mu) / sigma
X_test = (X_test - mu) / sigma
```

```python
model=Anomaly_Detection()
model.fit(X_train)
```

```python
model.select_threshold(X_cv,y_cv)
```

```python
y_pred = model.predict(X_test)
model.evaluate(y_test, y_pred)
```
